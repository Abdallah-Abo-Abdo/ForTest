```python
# ============================================
#  إعداد مسار ثابت للمشروع محلياً (في نفس مجلد النوتبوك)
# ============================================
import os

# os.getcwd() يعطينا مكان تشغيل النوتبوك الحالي
BASE_DIR = os.getcwd()

PROJECT_DIR = os.path.join(BASE_DIR, "Jupyter_Python_Basics")
DATA_DIR = os.path.join(PROJECT_DIR, "data")

os.makedirs(DATA_DIR, exist_ok=True)

print("✅ Base folder:", BASE_DIR)
print("✅ Project folder:", PROJECT_DIR)
print("✅ Data folder:", DATA_DIR)
print("📂 ملفات المشروع:", os.listdir(PROJECT_DIR))
```

```python
# ============================================
#  كتابة وقراءة ملف نصي (TXT) محلياً
# ============================================
import os

# نعيد استخدام DATA_DIR من الخلية السابقة
note_path = os.path.join(DATA_DIR, "notes.txt")

# كتابة ملف (سيبقى محفوظ على جهازك)
with open(note_path, "w", encoding="utf-8") as f:
    f.write("هذه ملاحظة محفوظة محلياً داخل جهازك (Jupyter).\n")
    f.write("إذا أغلقت Jupyter ورجعت لاحقاً ستجد الملف موجود.\n")

# قراءة الملف
with open(note_path, "r", encoding="utf-8") as f:
    content = f.read()

print("✅ تم الحفظ في:", note_path)
print("----- محتوى الملف -----")
print(content)
```
