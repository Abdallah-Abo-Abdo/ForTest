### 4.2 عمليات حسابية


```python
# ============================================
#  عمليات حسابية
# ============================================
a = 10
b = 3

print("a+b =", a+b)
print("a-b =", a-b)
print("a*b =", a*b)
print("a/b =", a/b)
print("a//b =", a//b)  # قسمة صحيحة
print("a%b =", a%b)    # باقي القسمة
print("a**b =", a**b)  # أس
```

### 4.5 شروط وحلقات


```python
# ============================================
#  if / for / while
# ============================================
x = 7

if x > 10:
    print("x أكبر من 10")
elif x == 10:
    print("x يساوي 10")
else:
    print("x أصغر من 10")

print("\nLoop for:")
for i in range(5):
    print("i =", i)

print("\nLoop while:")
c = 3
while c > 0:
    print("count =", c)
    c -= 1
```

### 4.6 دوال (Functions)


```python
# ============================================
# Functions
# ============================================
def calc_area(width, height):
    # هذه دالة: تستقبل قيمتين وتُرجع الناتج
    return width * height

print("Area(5,4) =", calc_area(5, 4))
print("Area(10,2) =", calc_area(10, 2))
```

## 6) ملفات CSV داخل Drive (حفظ وقراءة)
سننشئ ملف CSV داخل مجلد `data` في Drive.


```python
# ============================================
#    إنشاء CSV وقراءته (بدون pandas)
# ============================================
import csv

csv_path = os.path.join(DATA_DIR, "students.csv")

rows = [
    ["name", "course", "score"],
    ["Ali", "Python", 90],
    ["Maha", "Python", 95],
    ["Omar", "Python", 88],
]

# كتابة CSV
with open(csv_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print("✅ تم إنشاء CSV في:", csv_path)

# قراءة CSV
print("\n--- قراءة CSV ---")
with open(csv_path, "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## 7) حفظ بيانات كـ JSON داخل Drive
مفيد جداً لتخزين إعدادات/قوائم/معلومات.


```python
# ============================================
#    JSON حفظ/قراءة داخل Drive
# ============================================
import json

json_path = os.path.join(DATA_DIR, "settings.json")

settings = {
    "course": "Python Basics",
    "level": "Beginner",
    "city": "Amman",
    "features": ["files", "drive", "csv", "plots"]
}

# حفظ JSON
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(settings, f, ensure_ascii=False, indent=2)

print("✅ تم حفظ JSON في:", json_path)

# قراءة JSON
with open(json_path, "r", encoding="utf-8") as f:
    loaded = json.load(f)

print("✅ البيانات بعد القراءة:", loaded)
```

## 8) رسم بسيط (Matplotlib) + حفظ الرسم كصورة في Drive
Colab ممتاز لإظهار الرسومات، والأهم: تقدر تحفظها داخل Drive.


```python
# ============================================
#    رسم بسيط + حفظ PNG داخل Drive
# ============================================
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y, marker="o")
plt.title("Simple Plot in Colab")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()

plot_path = os.path.join(DATA_DIR, "plot.png")
plt.figure()
plt.plot(x, y, marker="o")
plt.title("Saved Plot")
plt.grid(True)
plt.savefig(plot_path, dpi=150)
plt.close()

print("✅ تم حفظ الرسم في:", plot_path)
```
